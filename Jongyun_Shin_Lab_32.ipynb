```
k=0.5
def f1(x):
  return k*x*(1-x)

def f2(x):
  return k*x*(1-x)**1.5

def f3(x):
  return k*x*(1-x)**0.5

def f4(x):
  return k*x*(1-x)**2

def f5(x):
  return k*x*(1-x)**0.75

```

## Q1
* Copy/paste your `equilibrium(f,a,n)` function from the prelab to a code cell and run it.  
* Use it to graph the first `100` terms of the recursive sequence defined by your function, with initial value `0.5271`.  (I chose this value because it is unlikely to be at an equilibrium by coincidence).  
* What equilibrium does your sequence appear to approach?  Check with your group and see that all of your functions exhibit the same behavior for `k=0.5`



```
def equilibrium(f, a, n):
    sequence = [a]
    for i in range(n):
        a = f(a)
        sequence.append(a)

    x = np.arange(0,1,0.001)
    y= f(x)

    fig, axs = plt.subplots(1,2,figsize=(10,5))

    axs[0].grid()
    axs[0].plot(x,y)
    axs[0].plot(x,x)
    axs[0].set_aspect('equal')
    axs[0].set_title('f(x) and y=x')


    axs[1].grid()
    axs[1].set_ylim([0,1])
    axs[1].plot(sequence,'o')
    axs[1].set_title('Recursive Sequence')

    plt.show()

    return sequence[-1]

print(equilibrium(f3, 0.5271, 100))
```

The two graphs meet at 0, converges to zero

Since the outcome of the sequences are oscillating between two values of numbers, we should modify 𝑎𝑛=𝑎𝑛+1 to 𝑎𝑛=𝑎𝑛+2, so that we could approach equilibrium to one number by observing every second number instead of the next.

```
def equilibrium_2(f, a, n):
    sequence = [a]
    for i in range(n):
        a+1 == f(a)
        sequence.append(a)

    x = np.arange(0,1,0.001)
    y= f(x)

    fig, axs = plt.subplots(1,2,figsize=(10,5))

    axs[0].grid()
    axs[0].plot(x,y)
    axs[0].plot(x,x)
    axs[0].set_aspect('equal')
    axs[0].set_title('f(x) and y=x')


    axs[1].grid()
    axs[1].set_ylim([0,1])
    axs[1].plot(sequence,'o')
    axs[1].set_title('Recursive Sequence')

    plt.show()

    return sequence[-1]
def f3_23(x):
  return 2.3*x*(1-x)**0.5

print(equilibrium_2(f3_23, 0.5271, 100))
```

If we repeat this for f'(x), we get the equation b=(3-k^2)/2 given that
f'(x) is 0.5(1-x)^0.5-1/4x(1-x)^-0.5.

To clean up, we first have the equation:
b=1-(1/k^2), which finds the fixed points,and: b=(3-k^2)/2, which simplifies what the derivative is at the fixed points
