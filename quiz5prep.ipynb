```python
#5. Create a new column of the DataFrame called "tokens" where each string in the reviews column is converted to a list of individual lower-case words.

df["tokens"] = df["review"].str.lower().str.split()
df.head()


#or 

def lower_list(lst):
    out = []

    # If there is a missing value (NaN) in the "review" column,
    # the function returns an empty list instead of failing.
    if not isinstance(lst, list):
        return out
    
    # Convert each string in the list to lowercase
    for w in lst:
        out.append(w.lower())
    
    return out


# Split into word lists, then convert to lowercase using apply()
df["tokens"] = df["review"].str.split().apply(lower_list)
df["tokens"].head()
```
