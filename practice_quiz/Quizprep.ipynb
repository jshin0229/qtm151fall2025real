4. Write a function that takes in a single number as an input argument, and returns a single string as the output. The value of the output string is one of two choices, which is determined by whether the input satisfies a logical condition. Then use the .apply() method from Pandas to evaluate this function for every row in a DataFrame. You will be told which column/variable should be input to the function.

 



```python
def mpg_label(value):
    if value >= 25:
        return "High"
    else:
        return "Low"
    
df["mpg_label"] = df["mpg"].apply(mpg_label)

print(df[["mpg_label", "mpg"]])
```

```python
# returns one of two strings based on a number (handle non-numeric horsepower)
# convert horsepower column to numeric (coerce non-numeric to NaN)
hp_numeric = pd.to_numeric(df["horsepower"], errors="coerce")

def tag_hp(value):
    if pd.isna(value):
        return "Unknown"   # missing or non-numeric horsepower
    if value >= 150:      # <-- change the condition/number if your prompt says so
        return "High"
    else:
        return "Low"

df["hp_tag"] = hp_numeric.apply(tag_hp)
print(df[["hp_tag", "horsepower"]])
```

5. Given a DataFrame, you will want to randomly assign each row to one of two groups. To do so, you will add a new column to the DataFrame called "random_var" which is filled with random numbers from a given distribution. Then you will create two new DataFrames, based on the whether each row's random variable satisfies a given logical condition. These two DataFrames represent the random assignment into two groups.

 


```python
import numpy as np
import pandas as pd

# read the dataset
df = pd.read_csv("practice_quiz_data/features.csv")

# get size of data-set
num_rows, num_cols = df.shape

# randomly generate a number (1 or 2) for each person in the dataset
group_no = np.random.choice([1, 2], size=num_rows)

# add this new column to the dataframe
df["GroupID"] = group_no

# check first few rows
print(df.head())

# (optional) print how many people are in each group
print(df["GroupID"].value_counts())
```

6. Continuing question 5, you will make two scatter plots from the data in each randomly assigned group. So that they appear side-by-side, you will need to use the subplots command from the matplotlib.pyplot library.

```python
import matplotlib.pyplot as plt

# split the data into two groups from Question 5
group1 = df.query("GroupID == 1")
group2 = df.query("GroupID == 2")

# create two subplots side by side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# scatter plot for Group 1
ax1.scatter(x=group1["weight"], y=group1["acceleration"], color="blue")
ax1.set_title("Group 1")
ax1.set_xlabel("Weight")
ax1.set_ylabel("Acceleration")

# scatter plot for Group 2
ax2.scatter(x=group2["weight"], y=group2["acceleration"], color="orange")
ax2.set_title("Group 2")
ax2.set_xlabel("Weight")
ax2.set_ylabel("Acceleration")

plt.tight_layout()
plt.show()
```
