You are given two lists of numbers, x_vals and y_vals, and must create a Numpy array where the components are related to those in the lists (for example, you might be asked to create a Numpy array with entries zk=(xk)**2-5*sin(yk)). Then generate a histogram of the values in the Numpy array, with labeled axes and a title. You will also be asked to change the color of the bars in the histogram.

```python
#practice_quiz-solutions, 

# 1) Your input lists (same length!)
x_vals = [1.0, 2.0, 3.0, 4.0]
y_vals = [0.1, 0.5, 0.9, 1.2]

# 2) Convert to NumPy arrays for vectorized math
x = np.array(x_vals)
y = np.array(y_vals)

# 3) Define the derived array (edit this formula as needed)
# Example: z_k = (x_k)**2 - 5*sin(y_k)
z = x**2 - 5*np.sin(y)

# 4) Plot a histogram
plt.hist(z, bins=20, color="red", edgecolor="black")
plt.xlabel("z values")
plt.ylabel("Frequency")
plt.title("Histogram of z = x^2 - 5 sin(y)")
plt.show()



```

3. Generate random samples from two specified distributions and plot their histograms, side by side in a single figure. Each sub-figure will have its own labeled axes and title.

```python
#assignment 3 

#random seed
np.random.seed(42)  

#distribution 1
sample_1 = np.random.normal(loc=50, scale=10, size=1000)
#distribution 2
sample_2 = np.random.exponential(scale=15, size=1000)

# Create subplots
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Plot histogram for sample_1
axes[0].hist(sample_1, bins=30, color='blue', edgecolor='black')
axes[0].set_title('Normal Distribution (mean=50, std=10)')
axes[0].set_xlabel('Value')
axes[0].set_ylabel('Frequency') 

#plot histogram for sample_2
axes[1].hist(sample_2, bins=30, color='green', edgecolor='black')
axes[1].set_title('Exponential Distribution (scale=15)')
axes[1].set_xlabel('Value')
axes[1].set_ylabel('Frequency')


plt.grid()
plt.show()

```

5. Read in a .csv file to create a Pandas DataFrame. Loop over a list of variable/column names, printing the mean of each variable.

```python
# imports (put these once at the top of your notebook)
import pandas as pd

# 1) Read the CSV into a DataFrame
df = pd.read_csv("practice_quiz_data/united_states_co2_emissions.csv")   # <-- replace with your path

# 2) Choose which columns to summarize (edit to your real column names)
cols = ["year", "total_emissions"]

# 3) Loop and print the mean of each column
for c in cols:
    if c not in df.columns:
        print(f"[skip] Column not found: {c}")
        continue

    # coerce to numeric (non-numeric → NaN), and compute mean ignoring NaNs
    col_numeric = pd.to_numeric(df[c], errors="coerce")
    mean_val = col_numeric.mean()
    print(f"{c}: {mean_val:.6g}")
```

```python
# List of CSV file names
csv_files = [
    "practice_quiz_data/china_co2_emissions.csv",
    "practice_quiz_data/united_states_co2_emissions.csv",
    "practice_quiz_data/japan_co2_emissions.csv",
    "practice_quiz_data/india_co2_emissions.csv"
]

# Empty list to store means
means = []

# Compute mean of 'total_emissions' for each file
for file in csv_files:
    df = pd.read_csv(file)
    mean_val = df["total_emissions"].mean()
    means.append(mean_val)

print(means)
```
