3. Read in a .csv file to create a DataFrame which contains conversion rates for Euros to U.S. dollars and British Pounds to U.S. dollars over time.
```python
df = pd.read_csv("currency_2017.csv")

# Convert time column to datetime
df["time"] = pd.to_datetime(df["time"])

# Compute daily growth rates
# diff() gives (today - yesterday)
# shift(1) gives yesterday's value

df["euro_grow"] = (df["eur_usd"].diff() / df["eur_usd"].shift(1)) * 100
df["gbp_grow"]  = (df["gbp_usd"].diff() / df["gbp_usd"].shift(1)) * 100

# Plot both growth series on one graph
df_delhi[   ["time","euro_grow","gbp_grow"]   ].set_index("time").plot()

plt.xlabel("Time")
plt.ylabel("Percent Growth (%)")
plt.title("Currency Growth Rates: EUR→USD and GBP→USD")
plt.show()
```
